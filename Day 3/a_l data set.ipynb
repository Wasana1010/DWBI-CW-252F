```python
# ============================================================
# GENERIC CSV DATASET PROFILER - GOOGLE COLAB VERSION
# ============================================================

import pandas as pd
import numpy as np
from google.colab import files
import io


# ------------------------------------------------------------
# 1. Upload CSV file
# ------------------------------------------------------------

print("Please upload your CSV dataset:")

uploaded = files.upload()

# Get uploaded file name automatically
file_name = next(iter(uploaded))

print("\nUploaded file:", file_name)


# ------------------------------------------------------------
# 2. Load dataset
# ------------------------------------------------------------

try:
    df = pd.read_csv(io.BytesIO(uploaded[file_name]))

except UnicodeDecodeError:
    # Try another common encoding
    df = pd.read_csv(
        io.BytesIO(uploaded[file_name]),
        encoding="latin1"
    )

print("\nDataset loaded successfully!")


# ------------------------------------------------------------
# 3. Basic Dataset Information
# ------------------------------------------------------------

print("\n========== DATASET PROFILE ==========\n")

print("File Name:")
print(file_name)

print("\nDataset Shape:")
print("Rows   :", df.shape[0])
print("Columns:", df.shape[1])


# ------------------------------------------------------------
# 4. Column Names
# ------------------------------------------------------------

print("\n========== COLUMN NAMES ==========\n")

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")


# ------------------------------------------------------------
# 5. Data Types
# ------------------------------------------------------------

print("\n========== DATA TYPES ==========\n")

print(df.dtypes)


# ------------------------------------------------------------
# 6. First 5 Rows
# ------------------------------------------------------------

print("\n========== FIRST 5 ROWS ==========\n")

print(df.head())


# ------------------------------------------------------------
# 7. Missing Values
# ------------------------------------------------------------

print("\n========== MISSING VALUES ==========\n")

missing_values = df.isnull().sum()

missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)

missing_report = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage (%)": missing_percentage
})

print(missing_report)


# ------------------------------------------------------------
# 8. Duplicate Rows
# ------------------------------------------------------------

print("\n========== DUPLICATE ROWS ==========\n")

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)


# ------------------------------------------------------------
# 9. Unique Values
# ------------------------------------------------------------

print("\n========== UNIQUE VALUES ==========\n")

unique_values = df.nunique()

print(unique_values)


# ------------------------------------------------------------
# 10. Numerical Column Statistics
# ------------------------------------------------------------

print("\n========== NUMERICAL STATISTICS ==========\n")

numeric_columns = df.select_dtypes(
    include=np.number
).columns

if len(numeric_columns) > 0:
    print(df[numeric_columns].describe())
else:
    print("No numerical columns found.")


# ------------------------------------------------------------
# 11. Categorical Column Statistics
# ------------------------------------------------------------

print("\n========== CATEGORICAL STATISTICS ==========\n")

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

if len(categorical_columns) > 0:
    print(df[categorical_columns].describe())
else:
    print("No categorical columns found.")


# ------------------------------------------------------------
# 12. Detailed Column Profile
# ------------------------------------------------------------

print("\n========== DETAILED COLUMN PROFILE ==========\n")

profile_data = []

for column in df.columns:

    total_values = len(df[column])

    missing = df[column].isnull().sum()

    unique = df[column].nunique()

    dtype = str(df[column].dtype)

    profile_data.append({
        "Column Name": column,
        "Data Type": dtype,
        "Total Values": total_values,
        "Missing Values": missing,
        "Missing %": round(
            (missing / total_values) * 100, 2
        ) if total_values > 0 else 0,
        "Unique Values": unique
    })


profile_df = pd.DataFrame(profile_data)

print(profile_df)


# ------------------------------------------------------------
# 13. Memory Usage
# ------------------------------------------------------------

print("\n========== MEMORY USAGE ==========\n")

memory_usage = df.memory_usage(deep=True).sum()

print(
    "Total Memory Usage:",
    round(memory_usage / 1024, 2),
    "KB"
)


# ------------------------------------------------------------
# 14. Save Dataset Profile to CSV
# ------------------------------------------------------------

output_file = "dataset_profile.csv"

profile_df.to_csv(
    output_file,
    index=False
)

print("\nDataset profile saved as:", output_file)


# ------------------------------------------------------------
# 15. Download Profile
# ------------------------------------------------------------

files.download(output_file)


print("\n========== PROFILING COMPLETED ==========")
```

### Data Cleaning and Transformation

This step addresses the identified issues:
1.  **Numeric Conversion for Zscore**: Converts the `Zscore` column to a numeric type, coercing any non-numeric values to `NaN`.
2.  **Standardize Categorical Columns**: Converts relevant categorical columns to lowercase to handle inconsistent capitalization.
3.  **Handle Missing Values**: Removes rows containing any `NaN` values, ensuring a clean dataset for analysis.

```python
# Make a copy to avoid modifying the original 'df' if not intended for broader changes
df_transformed = df.copy()

# 1. Convert 'Zscore' to numeric, coercing errors to NaN
df_transformed['Zscore'] = pd.to_numeric(df_transformed['Zscore'], errors='coerce')

# 2. Identify categorical columns that might have capitalization issues and convert to lowercase
# Based on df.dtypes and common categorical columns in this dataset
string_cols_to_lower = ['stream', 'sub1', 'sub1_r', 'sub2', 'sub2_r', 'sub3', 'sub3_r', 'cgt_r', 'ge_r', 'syllabus', 'gender']

for col in string_cols_to_lower:
    if col in df_transformed.columns:
        df_transformed[col] = df_transformed[col].astype(str).str.lower()

# Redefine biological_science_high_zscore_students after initial transformations
biological_science_high_zscore_students_transformed = df_transformed[
    (df_transformed['stream'] == 'biological science') &
    (df_transformed['Zscore'] > 0.4)
].copy() # Use .copy() to avoid SettingWithCopyWarning

# 3. Drop rows with any missing values
initial_rows_transformed = len(biological_science_high_zscore_students_transformed)
biological_science_high_zscore_students_final = biological_science_high_zscore_students_transformed.dropna()
rows_after_cleaning_final = len(biological_science_high_zscore_students_final)

print(f"Original number of rows for biological_science_high_zscore_students after initial transform: {initial_rows_transformed}")
print(f"Number of rows after dropping nulls: {rows_after_cleaning_final}")
print(f"Number of rows dropped due to nulls: {initial_rows_transformed - rows_after_cleaning_final}")

display(biological_science_high_zscore_students_final.head())
print("\nInfo of the final DataFrame after transformations:")
biosci_df_info = io.StringIO()
biological_science_high_zscore_students_final.info(buf=biosci_df_info)
print(biosci_df_info.getvalue())
```
