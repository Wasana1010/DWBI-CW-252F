### Data Cleaning and Transformation

This step addresses the identified issues:
1.  **Numeric Conversion for Zscore**: Converts the `Zscore` column to a numeric type, coercing any non-numeric values to `NaN`.
2.  **Standardize Categorical Columns**: Converts relevant categorical columns to lowercase to handle inconsistent capitalization.
3.  **Handle Missing Values**: Removes rows containing any `NaN` values, ensuring a clean dataset for analysis.

```python
# Make a copy to avoid modifying the original 'df' if not intended for broader changes
df_transformed = df.copy()

# 1. Convert 'Zscore' to numeric, coercing errors to NaN
df_transformed['Zscore'] = pd.to_numeric(df_transformed['Zscore'], errors='coerce')

# 2. Identify categorical columns that might have capitalization issues and convert to lowercase
# Based on df.dtypes and common categorical columns in this dataset
string_cols_to_lower = ['stream', 'sub1', 'sub1_r', 'sub2', 'sub2_r', 'sub3', 'sub3_r', 'cgt_r', 'ge_r', 'syllabus', 'gender']

for col in string_cols_to_lower:
    if col in df_transformed.columns:
        df_transformed[col] = df_transformed[col].astype(str).str.lower()

# Redefine biological_science_high_zscore_students after initial transformations
biological_science_high_zscore_students_transformed = df_transformed[
    (df_transformed['stream'] == 'biological science') &
    (df_transformed['Zscore'] > 0.4)
].copy() # Use .copy() to avoid SettingWithCopyWarning

# 3. Drop rows with any missing values
initial_rows_transformed = len(biological_science_high_zscore_students_transformed)
biological_science_high_zscore_students_final = biological_science_high_zscore_students_transformed.dropna()
rows_after_cleaning_final = len(biological_science_high_zscore_students_final)

print(f"Original number of rows for biological_science_high_zscore_students after initial transform: {initial_rows_transformed}")
print(f"Number of rows after dropping nulls: {rows_after_cleaning_final}")
print(f"Number of rows dropped due to nulls: {initial_rows_transformed - rows_after_cleaning_final}")

display(biological_science_high_zscore_students_final.head())
print("\nInfo of the final DataFrame after transformations:")
biosci_df_info = io.StringIO()
biological_science_high_zscore_students_final.info(buf=biosci_df_info)
print(biosci_df_info.getvalue())
```

```python
from google.colab import files
import pandas as pd

# This will open the file picker
uploaded = files.upload()

# Automatically use whatever CSV you select
file_name = next(iter(uploaded))

print("Uploaded:", file_name)

olap_data = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Shape:", olap_data.shape)

display(olap_data.head())
```

```python
# Prepare numeric measures for OLAP

olap_data['Zscore'] = pd.to_numeric(
    olap_data['Zscore'],
    errors='coerce'
)

olap_data['district_rank_num'] = (
    olap_data['district_rank']
    .astype(str)
    .str.extract(r'(\d+)')[0]
)

olap_data['district_rank_num'] = pd.to_numeric(
    olap_data['district_rank_num'],
    errors='coerce'
)

olap_data['island_rank_num'] = (
    olap_data['island_rank']
    .astype(str)
    .str.extract(r'(\d+)')[0]
)

olap_data['island_rank_num'] = pd.to_numeric(
    olap_data['island_rank_num'],
    errors='coerce'
)

olap_data['cgt_score'] = pd.to_numeric(
    olap_data['cgt_r'],
    errors='coerce'
)

print("OLAP measures prepared successfully!")

display(
    olap_data[
        ['Zscore', 'district_rank_num', 'island_rank_num', 'cgt_score']
    ].head()
)
```

```python
# ============================================================
# DIMENSION TABLE 1: STUDENT
# ============================================================

dim_student = olap_data[
    [
        'index',
        'gender',
        'birth_day',
        'birth_month',
        'birth_year'
    ]
].drop_duplicates().reset_index(drop=True)

# Create surrogate key
dim_student.insert(
    0,
    'student_key',
    range(1, len(dim_student) + 1)
)

# Rename original index column
dim_student.rename(
    columns={'index': 'source_index'},
    inplace=True
)

print("DIM_STUDENT created successfully!")
print("Shape:", dim_student.shape)

display(dim_student.head())
```

```python
# ============================================================
# DIMENSION TABLE 2: EXAM
# ============================================================

dim_exam = olap_data[
    [
        'al_year',
        'stream',
        'syllabus'
    ]
].drop_duplicates().reset_index(drop=True)

# Create surrogate key
dim_exam.insert(
    0,
    'exam_key',
    range(1, len(dim_exam) + 1)
)

print("DIM_EXAM created successfully!")
print("Shape:", dim_exam.shape)

display(dim_exam)
```

```python
# ============================================================
# DIMENSION TABLE 3: SUBJECT
# ============================================================

dim_subject = olap_data[
    [
        'sub1',
        'sub2',
        'sub3'
    ]
].drop_duplicates().reset_index(drop=True)

# Create surrogate key
dim_subject.insert(
    0,
    'subject_key',
    range(1, len(dim_subject) + 1)
)

# Rename columns
dim_subject.rename(
    columns={
        'sub1': 'subject_1',
        'sub2': 'subject_2',
        'sub3': 'subject_3'
    },
    inplace=True
)

print("DIM_SUBJECT created successfully!")
print("Shape:", dim_subject.shape)

display(dim_subject)
```

```python
# ============================================================
# FACT TABLE: FACT_RESULTS
# ============================================================

fact = olap_data.copy()

# 1. Connect DIM_STUDENT
fact = fact.merge(
    dim_student[['student_key', 'source_index']],
    left_on='index',
    right_on='source_index',
    how='left'
)

# 2. Connect DIM_EXAM
fact = fact.merge(
    dim_exam,
    on=['al_year', 'stream', 'syllabus'],
    how='left'
)

# 3. Prepare DIM_SUBJECT for connection
subject_lookup = dim_subject.rename(
    columns={
        'subject_1': 'sub1',
        'subject_2': 'sub2',
        'subject_3': 'sub3'
    }
)

# Connect DIM_SUBJECT
fact = fact.merge(
    subject_lookup,
    on=['sub1', 'sub2', 'sub3'],
    how='left'
)

# 4. Create final FACT table
fact_results = fact[
    [
        'student_key',
        'exam_key',
        'subject_key',
        'Zscore',
        'district_rank_num',
        'island_rank_num',
        'cgt_score',
        'sub1_r',
        'sub2_r',
        'sub3_r',
        'ge_r'
    ]
].copy()

# Create Fact Table primary key
fact_results.insert(
    0,
    'result_key',
    range(1, len(fact_results) + 1)
)

# Rename columns
fact_results.rename(
    columns={
        'Zscore': 'zscore',
        'district_rank_num': 'district_rank',
        'island_rank_num': 'island_rank',
        'sub1_r': 'subject_1_grade',
        'sub2_r': 'subject_2_grade',
        'sub3_r': 'subject_3_grade',
        'ge_r': 'general_english_grade'
    },
    inplace=True
)

print("FACT_RESULTS created successfully!")
print("Shape:", fact_results.shape)

display(fact_results.head())
```

```python
# ============================================================
# FINAL STAR SCHEMA VALIDATION
# ============================================================

print("=" * 60)
print("STAR SCHEMA VALIDATION")
print("=" * 60)

print("\nDimension Tables")
print("DIM_STUDENT :", dim_student.shape)
print("DIM_EXAM    :", dim_exam.shape)
print("DIM_SUBJECT :", dim_subject.shape)

print("\nFact Table")
print("FACT_RESULTS:", fact_results.shape)

# Check missing foreign keys
print("\nMissing Foreign Keys")

print(
    "Missing student_key:",
    fact_results['student_key'].isna().sum()
)

print(
    "Missing exam_key:",
    fact_results['exam_key'].isna().sum()
)

print(
    "Missing subject_key:",
    fact_results['subject_key'].isna().sum()
)

# Preview tables
print("\nDIM_STUDENT")
display(dim_student.head())

print("\nDIM_EXAM")
display(dim_exam.head())

print("\nDIM_SUBJECT")
display(dim_subject.head())

print("\nFACT_RESULTS")
display(fact_results.head())
```

```python
# ============================================================
# 5. POPULATE OLAP CUBE
# ============================================================

olap_cube_data = (
    fact_results
    .merge(dim_student, on='student_key', how='left')
    .merge(dim_exam, on='exam_key', how='left')
    .merge(dim_subject, on='subject_key', how='left')
)

print("OLAP cube data populated successfully!")
print("Shape:", olap_cube_data.shape)

display(olap_cube_data.head())
```

```python
# ============================================================
# 6. VALIDATE AND TEST OLAP CUBE
# ============================================================

print("=" * 60)
print("OLAP CUBE VALIDATION")
print("=" * 60)

# 1. Check cube size
print("\nCube Shape:")
print(olap_cube_data.shape)

# 2. Check missing dimension keys
print("\nMissing Keys:")
print("student_key :", olap_cube_data['student_key'].isna().sum())
print("exam_key    :", olap_cube_data['exam_key'].isna().sum())
print("subject_key :", olap_cube_data['subject_key'].isna().sum())

# 3. Check important measures
print("\nMeasure Validation:")
print("Missing Z-score :", olap_cube_data['zscore'].isna().sum())
print("Missing CGT     :", olap_cube_data['cgt_score'].isna().sum())

# 4. Test aggregation
test_cube = olap_cube_data.groupby('gender').agg(
    student_count=('student_key', 'count'),
    average_zscore=('zscore', 'mean'),
    min_zscore=('zscore', 'min'),
    max_zscore=('zscore', 'max')
).reset_index()

print("\nTest Cube - Performance by Gender:")
display(test_cube)

# 5. Check total records
print("\nRecord Validation:")
print("Fact table records :", len(fact_results))
print("Cube records       :", len(olap_cube_data))

if len(fact_results) == len(olap_cube_data):
    print("Record count validation: PASSED")
else:
    print("Record count validation: FAILED")
```
